```python
import geopandas as gpd
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error

# Load shapefile
shapefile_path = "fold_25k_ngdr_20250224140917945.shp"
gdf = gpd.read_file(shapefile_path)

print("✅ Initial columns:", gdf.columns.tolist())
print(gdf.head())

# Drop geometry column if it exists
if "geometry" in gdf.columns:
    gdf = gdf.drop(columns=["geometry"])

# Drop columns with all missing or only one unique value
for col in gdf.columns:
    if gdf[col].nunique() <= 1 or gdf[col].isnull().sum() == len(gdf):
        gdf.drop(columns=[col], inplace=True)

# Drop known irrelevant columns
irrelevant_columns = ["accession_", "remarks", "objectid", "id"]  # Adjust based on actual dataset
gdf = gdf.drop(columns=[col for col in irrelevant_columns if col in gdf.columns], errors="ignore")

# Handle missing values
for col in gdf.columns:
    if gdf[col].dtype == "object":
        gdf[col] = gdf[col].fillna("Unknown")
    else:
        gdf[col] = gdf[col].fillna(gdf[col].mean())

# Encode categorical variables
for col in gdf.select_dtypes(include=["object"]).columns:
    le = LabelEncoder()
    gdf[col] = le.fit_transform(gdf[col].astype(str))

# Correlation-based feature filtering (optional thresholding)
correlation_matrix = gdf.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()

# Set target variable
target_column = "shape_leng" if "shape_leng" in gdf.columns else "dip_amount"
if target_column not in gdf.columns:
    target_column = gdf.columns[-1]  # fallback

# Drop low-correlation features (optional: abs(corr) < 0.05 with target)
low_corr_features = correlation_matrix[target_column].abs()[correlation_matrix[target_column].abs() < 0.05].index.tolist()
gdf = gdf.drop(columns=low_corr_features, errors="ignore")

# Separate features and target
X = gdf.drop(columns=[target_column])
y = gdf[target_column]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Define and train models
models = {
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=6, objective="reg:squarederror", random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=6, random_state=42)
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    accuracy = 100 * (1 - (rmse / np.mean(y_test)))
    results[name] = accuracy
    print(f"{name} Accuracy: {accuracy:.2f}%")

# Best model
best_model_name = max(results, key=results.get)
print(f"\n🏆 Best Model: {best_model_name} with Accuracy: {results[best_model_name]:.2f}%")
```

```python
import geopandas as gpd
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error

# Load shapefile
shapefile_path = "fold_25k_ngdr_20250224140917945.shp"
gdf = gpd.read_file(shapefile_path)

print("✅ Initial columns:", gdf.columns.tolist())
print(gdf.head())

# Drop geometry column if it exists
gdf = gdf.drop(columns=["geometry"], errors="ignore")

# Drop columns with all missing or only one unique value
gdf = gdf.dropna(axis=1, how='all')  # Drop columns where all values are NaN
gdf = gdf.loc[:, gdf.nunique() > 1]  # Drop columns with only one unique value

# Drop known irrelevant columns
irrelevant_columns = ["accession_", "remarks", "objectid", "id"]  # Adjust based on actual dataset
gdf = gdf.drop(columns=[col for col in irrelevant_columns if col in gdf.columns], errors="ignore")

# Handle missing values
gdf.fillna(method='ffill', inplace=True)  # Forward fill missing values
gdf.fillna(method='bfill', inplace=True)  # Backward fill if necessary

gdf = gdf.apply(lambda x: x.fillna(x.mean()) if x.dtype != 'O' else x.fillna("Unknown"))

# Encode categorical variables
for col in gdf.select_dtypes(include=["object"]).columns:
    le = LabelEncoder()
    gdf[col] = le.fit_transform(gdf[col].astype(str))

# Correlation-based feature filtering (optional thresholding)
correlation_matrix = gdf.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()

# Set target variable
target_column = "shape_leng" if "shape_leng" in gdf.columns else "dip_amount"
if target_column not in gdf.columns:
    target_column = gdf.columns[-1]  # fallback

# Drop low-correlation features (optional: abs(corr) < 0.05 with target)
low_corr_features = correlation_matrix[target_column].abs()[correlation_matrix[target_column].abs() < 0.05].index.tolist()
gdf = gdf.drop(columns=low_corr_features, errors="ignore")

# Separate features and target
X = gdf.drop(columns=[target_column])
y = gdf[target_column]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Define and train models
models = {
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=6, objective="reg:squarederror", random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=6, random_state=42)
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    accuracy = 100 * (1 - (rmse / np.mean(y_test)))
    results[name] = accuracy
    print(f"{name} Accuracy: {accuracy:.2f}%")

# Best model
best_model_name = max(results, key=results.get)
print(f"\n🏆 Best Model: {best_model_name} with Accuracy: {results[best_model_name]:.2f}%")
```
